```python
taxids = list(name2taxid.values())
taxids
```

```python
# metadata mapping ffrom TaxID to GTDB name
metadata = pd.read_csv('bac120_metadata_r226.tsv', sep='\t') # file downloaded from GTDB release r226 https://data.gtdb.aau.ecogenomic.org/releases/release226/
metadata.head(2)
```

```python
gtdb_ids = []
for t in taxids:
    try:
        gtdb_ids.append(metadata.gtdb_taxonomy[metadata.ncbi_taxid==int(t)].values[0].split(';')[-1])
    except:
        print(t)
```
