```python
import sys
import pandas as pd
import numpy as np
from pathlib import Path
from tqdm import tqdm 
```

```python
# Define the root path
root = Path("path_to_chembl_data")  # Replace with the actual path to your ChEMBL data

# Files downloaded following instructions in README
csv_files = root.glob("*/joined_data_binarization/joined_all_max_activity.csv")
csv_files = list(csv_files)
csv_files = [str(f) for f in csv_files]
```

```python
len(csv_files)
```

```python
joined_df = pd.DataFrame()
for f in csv_files:
    df = pd.read_csv(f)
    df['species'] = str(f).split('/')[-3]
    joined_df = pd.concat([joined_df, df], ignore_index=True)
joined_df.shape
```
